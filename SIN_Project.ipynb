```python
# photosCLEFparsefinal.py

import xml.etree.ElementTree as ET
import pandas as pd

def parse_photos_CLEF(file_name='photosCLEF.xml'):
    """
    Parse photosCLEF.xml and extract:
      - photo_id, dateuploaded
      - latitude, longitude (from <location>)
      - title, description
      - owner_username, user_id (nsid), owner_location
    Returns a pandas DataFrame.
    """
    tree = ET.parse(file_name)
    root = tree.getroot()

    data = []
    for photo in root.findall('photo'):
        # Basic attributes
        photo_id     = photo.get('id')
        dateuploaded = photo.get('dateuploaded')

        # Geo‐tags under <location>
        loc_elem = photo.find('location')
        if loc_elem is not None:
            lat_str  = loc_elem.get('latitude')
            lon_str  = loc_elem.get('longitude')
            latitude  = float(lat_str)  if lat_str  is not None else None
            longitude = float(lon_str)  if lon_str  is not None else None
        else:
            latitude = None
            longitude = None

        # Title and description
        title_elem = photo.find('title')
        desc_elem  = photo.find('description')
        title       = title_elem.text      if title_elem is not None else None
        description = desc_elem.text       if desc_elem  is not None else None

        # Owner info: username, nsid (user_id), location
        owner_elem     = photo.find('owner')
        if owner_elem is not None:
            username = owner_elem.get('username')
            nsid     = owner_elem.get('nsid')
            location = owner_elem.get('location')
        else:
            username = None
            nsid     = None
            location = None

        data.append({
            'photo_id'      : photo_id,
            'dateuploaded'  : dateuploaded,
            'latitude'      : latitude,
            'longitude'     : longitude,
            'title'         : title,
            'description'   : description,
            'owner_username': username,
            'user_id'       : nsid,
            'owner_location': location
        })

    df = pd.DataFrame(data)

    # Convert dateuploaded from epoch seconds to datetime
    if df['dateuploaded'].notnull().all():
        df['dateuploaded'] = pd.to_datetime(
            pd.to_numeric(df['dateuploaded'], errors='coerce'),
            unit='s', errors='coerce'
        )

    return df

if __name__ == '__main__':
    # Ensure pandas will show all columns
    pd.set_option('display.max_columns', None)

    # Parse and load the DataFrame
    df = parse_photos_CLEF('photosCLEF.xml')

    # Specify columns to display (including the new user_id)
    cols = [
        'photo_id', 'dateuploaded',
        'latitude', 'longitude',
        'title', 'description',
        'owner_username', 'user_id', 'owner_location'
    ]

    print("Extracted photo records from photosCLEF.xml:")
    print(df[cols].head(10))
    print(f"\nTotal records: {len(df)}")
```

```python
import pandas as pd
import math
from IPython.display import display

# 1) Load & filter to only geo‑tagged photos
df = pd.read_csv('merged_photos.csv', parse_dates=['dateuploaded'])
df = df.dropna(subset=['dateuploaded','latitude','longitude'])
df = df.sort_values('dateuploaded').reset_index(drop=True)
print(f"Using {len(df)} geotagged photos (of 29195 total)")

# 2) Inline Haversine function (returns km)
def haversine_km(c1, c2):
    # c1, c2 = (lat, lon) in decimal degrees
    lat1, lon1 = math.radians(c1[0]), math.radians(c1[1])
    lat2, lon2 = math.radians(c2[0]), math.radians(c2[1])
    dlat, dlon = lat2 - lat1, lon2 - lon1
    a = (math.sin(dlat/2)**2 +
         math.cos(lat1) * math.cos(lat2) * math.sin(dlon/2)**2)
    return 6371.0 * 2 * math.asin(math.sqrt(a))

# 3) Sliding‐window co‑occurrence counter
def count_pairs(delta_t, delta_d_km):
    pairs = {}
    start = 0
    n = len(df)
    for i in range(n):
        t_i = df.at[i, 'dateuploaded']
        # advance window start so that dateuploaded[j] ≥ t_i - delta_t
        while df.at[start, 'dateuploaded'] < t_i - delta_t:
            start += 1
        for j in range(start, i):
            if df.at[j, 'user_id'] == df.at[i, 'user_id']:
                continue
            u, v = df.at[i,'user_id'], df.at[j,'user_id']
            pair = tuple(sorted((u, v)))
            coord_i = (df.at[i,'latitude'], df.at[i,'longitude'])
            coord_j = (df.at[j,'latitude'], df.at[j,'longitude'])
            if haversine_km(coord_i, coord_j) <= delta_d_km:
                pairs[pair] = pairs.get(pair, 0) + 1
    return pairs

# 4) Debug counts at different distance thresholds
delta_t = pd.Timedelta(hours=1)
print("\nTesting co‑occurrences at various distance thresholds:")
for d in [0.1, 0.5, 1.0]:  # 100 m, 500 m, 1 km
    p = count_pairs(delta_t, d)
    print(f"  Δd = {d} km → {len(p)} user‑pairs")

# 5) Pick the threshold that yields >0 (e.g., 0.5 km) and save results
chosen_d = 100
pairs = count_pairs(delta_t, chosen_d)

coocc_df = pd.DataFrame([
    {'user_u': u, 'user_v': v, 'cooccurrence_count': cnt}
    for (u, v), cnt in pairs.items()
])

# Save to CSV
coocc_df.to_csv('user_cooccurrences.csv', index=False)

# Make sure pandas will show all rows
import pandas as pd
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

# Display every co‑occurring pair
print(f"Found {len(coocc_df)} co‑occurring pairs at Δd={chosen_d} km, Δt=1 h:")
display(coocc_df)
```

```python
import pandas as pd
import math
import networkx as nx
import numpy as np
import random
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, f1_score, classification_report
from IPython.display import display

# 1) Load & filter your merged photo DataFrame
df = pd.read_pickle('df_all.pkl')  # or use CSV: pd.read_csv('merged_photos.csv', parse_dates=['dateuploaded'])
df = df.dropna(subset=['dateuploaded', 'latitude', 'longitude'])
df['user_id'] = df['user_id'].astype(str)
df = df.sort_values('dateuploaded').reset_index(drop=True)
print(f"Using {len(df)} geotagged photos")

# 2) Inline Haversine function (km)
def haversine_km(c1, c2):
    lat1, lon1 = math.radians(c1[0]), math.radians(c1[1])
    lat2, lon2 = math.radians(c2[0]), math.radians(c2[1])
    dlat, dlon = lat2 - lat1, lon2 - lon1
    a = math.sin(dlat / 2) ** 2 + math.cos(lat1) * math.cos(lat2) * math.sin(dlon / 2) ** 2
    return 6371.0 * 2 * math.asin(math.sqrt(a))

# 3) Count spatio‑temporal co‑occurrences
def count_pairs(delta_t, delta_d_km):
    pairs = {}
    start = 0
    n = len(df)
    for i in range(n):
        t_i = df.at[i, 'dateuploaded']
        while df.at[start, 'dateuploaded'] < t_i - delta_t:
            start += 1
        for j in range(start, i):
            u = str(df.at[i, 'user_id'])
            v = str(df.at[j, 'user_id'])
            if u == v:
                continue
            pair = tuple(sorted((u, v)))
            coord_i = (df.at[i, 'latitude'], df.at[i, 'longitude'])
            coord_j = (df.at[j, 'latitude'], df.at[j, 'longitude'])
            if haversine_km(coord_i, coord_j) <= delta_d_km:
                pairs[pair] = pairs.get(pair, 0) + 1
    return pairs

delta_t = pd.Timedelta(hours=1)
chosen_d = 100  # km threshold
pairs = count_pairs(delta_t, chosen_d)

coocc_df = pd.DataFrame([
    {'user_u': u, 'user_v': v, 'cooccurrence_count': cnt}
    for (u, v), cnt in pairs.items()
])
coocc_df.to_csv('user_cooccurrences.csv', index=False)
print(f"Saved {len(coocc_df)} co‑occurring pairs (Δd={chosen_d} km, Δt=1h)")
display(coocc_df)

# 4) Load Flickr friendship edges and cast to string
edges_df = pd.read_csv('flickrEdges.txt', sep='\t', header=None, names=['u', 'v'])
edges_df['u'] = edges_df['u'].astype(str)
edges_df['v'] = edges_df['v'].astype(str)
G = nx.Graph()
G.add_edges_from(edges_df.values)
print(f"\nFriendship graph: {G.number_of_nodes()} nodes, {G.number_of_edges()} edges")

# 5) Prepare examples for link prediction
pos = list(G.edges())
nodes = list(G.nodes())
neg = []
while len(neg) < len(pos):
    u, v = random.sample(nodes, 2)
    if not G.has_edge(u, v):
        neg.append((u, v))
pairs_all = pos + neg
labels = [1] * len(pos) + [0] * len(neg)

# 6) Build feature matrix (CN, Jaccard, AA, PA, cooccurrence)
coocc_dict = {
    tuple(sorted((str(r.user_u), str(r.user_v)))): r.cooccurrence_count
    for r in coocc_df.itertuples()
}

# Convert pairs to str and prepare labels
pairs_all = [(str(u), str(v)) for u, v in pos + neg]
labels = [1]*len(pos) + [0]*len(neg)

# Precompute features
cn_dict = {pair: len(list(nx.common_neighbors(G, *pair))) for pair in pairs_all}
aa_scores = {(u, v): p for u, v, p in nx.adamic_adar_index(G, pairs_all)}
pa_scores = {(u, v): p for u, v, p in nx.preferential_attachment(G, pairs_all)}
co_scores = {tuple(sorted((u, v))): coocc_dict.get(tuple(sorted((u, v))), 0)
             for u, v in pairs_all}

# Build final feature matrix
X = np.array([
    [cn_dict.get((u, v), 0),
     aa_scores.get((u, v), 0),
     pa_scores.get((u, v), 0),
     co_scores.get((u, v), 0)]
    for u, v in pairs_all
])
y = np.array(labels)

# 7) Train/test split & classifier
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=42
)
clf = LogisticRegression(max_iter=1000).fit(X_train, y_train)

# 8) Evaluate
y_pred = clf.predict(X_test)
y_prob = clf.predict_proba(X_test)[:, 1]
print("\n--- Evaluation Metrics ---")
print("ROC‑AUC:", roc_auc_score(y_test, y_prob))
print("F1 Score:", f1_score(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
import pandas as pd
import math

# Load photo metadata
df = pd.read_csv("merged_photos.csv", parse_dates=["dateuploaded"])
df = df.dropna(subset=['latitude', 'longitude', 'user_id'])
df['user_id'] = df['user_id'].astype(str)

# Create user_id → [(lat, lon), ...] mapping
user_locations = df.groupby('user_id')[['latitude', 'longitude']].apply(
    lambda x: list(x.itertuples(index=False))
).to_dict()

# Load co-occurrence data
coocc_df = pd.read_csv("user_cooccurrences.csv")
coocc_dict = {
    tuple(sorted((str(row.user_u), str(row.user_v)))): row.cooccurrence_count
    for row in coocc_df.itertuples()
}

# Haversine distance (km)
def haversine_km(c1, c2):
    lat1, lon1 = math.radians(c1[0]), math.radians(c1[1])
    lat2, lon2 = math.radians(c2[0]), math.radians(c2[1])
    dlat, dlon = lat2 - lat1, lon2 - lon1
    a = math.sin(dlat/2)**2 + math.cos(lat1)*math.cos(lat2)*math.sin(dlon/2)**2
    return 6371.0 * 2 * math.asin(math.sqrt(a))

# Suggest co-occurring users (with or without existing connections)
def suggest_friends_include_all(user_id, coocc_dict, user_locations, top_n=5):
    user_id = str(user_id)
    suggestions = []

    for (u, v), count in coocc_dict.items():
        if user_id not in (u, v):
            continue
        other = v if u == user_id else u

        coords_u = user_locations.get(user_id, [])
        coords_v = user_locations.get(other, [])
        if not coords_u or not coords_v:
            continue

        total_dist = 0
        num = 0
        for c1 in coords_u:
            for c2 in coords_v:
                total_dist += haversine_km(c1, c2)
                num += 1
        avg_dist = total_dist / num if num > 0 else float('inf')
        suggestions.append((other, count, avg_dist))

    ranked = sorted(suggestions, key=lambda x: (-x[1], x[2]))[:top_n]
    return pd.DataFrame(ranked, columns=["Suggested User", "Co-occurrence Count", "Avg Distance (km)"])

# ▶️ Try it with a sample user from your data
sample_user = "8088927@N06"  # Replace with any real user_id from your dataset

suggested_df = suggest_friends_include_all(sample_user, coocc_dict, user_locations, top_n=5)
print(f"\n Top Friend Suggestions for {sample_user}")
print(suggested_df)
```
